```python
import requests
import base64
import io
from PIL import Image
from IPython.display import display, Image as IPImage
```

```python
response = requests.get("http://127.0.0.1:8000")
print(response.content)
```

```python
json = {
    "link": "https://sample-videos.com/video321/mp4/720/big_buck_bunny_720p_2mb.mp4"
}

response = requests.post("http://127.0.0.1:8000/get_thumbnail", json=json)

print(response)
```

```python
thumbnail = response.json()["thumbnail"]

thumbnail = base64.b64decode(thumbnail)
```

```python
thumbnail = io.BytesIO(thumbnail).read()

with open("test.gif", "wb") as file:
    file.write(thumbnail)
```

```python
gif_path = "test.gif"
with Image.open(gif_path) as img:
    display(IPImage(filename=gif_path))
```
